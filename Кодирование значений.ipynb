# README

## README.md
Набор данных базы данных о мировом неравенстве
---------------------------------

## Содержимое этой папки

Помимо текущего файла README, эта папка содержит следующие CSV-файлы:

* "WID_countries.csv", содержащий список и значение всех кодов стран и регионов.

* Файл(ы) с именем "WID_data_XX.csv", содержащий данные для страны/региона XX.

* Файл(ы) с именем "WID_metadata_XX.csv", содержащий метаданные для страны/региона XX.

## Структура и формат CSV-файлов

В CSV-файлах в качестве разделителя используется точка с запятой ";". При необходимости строки заключаются в кавычки. Первая строка соответствует именам переменных.

### Структура файла WID_countries.csv

Файл WID_countries.csv содержит пять переменных:

* alpha2: двухбуквенный код страны/региона. В основном он соответствует номенклатуре ISO 3166-1 alpha-2 с некоторыми дополнениями для учета бывших стран, регионов и субрегионов. Регионы внутри страны XX обозначены как XX-YY, где YY — код региона. Регионы мира обозначены как XX и XX-MER, первый из которых использует паритет покупательной способности (по умолчанию), а второй — рыночные обменные курсы. Подробнее см. [техническое примечание «Цены и конвертация валют в WID.world»](https://wid.world/document/convert-wid-world-series/).

* titlename: название страны/региона, как оно будет представлено в предложении на английском языке (т.е. включая определенный артикль, если он есть).

* shortname: название страны/региона, как оно будет представлено отдельно на английском языке (т.е. без определенного артикля).

* region: регион мира, к которому принадлежит страна (аналогично разделению первого уровня геосхемы ООН).

* region2: подробный регион мира, к которому относится страна (аналогично разделению второго уровня геосхемы ООН).

### Структура файлов WID_data_XX.csv

Файлы WID_data_XX.csv содержат семь переменных:

* country: код страны/региона (см. WID_countries.csv).

* variable: код переменной WID (подробнее см. ниже).

* percentile: код процентиля WID (подробнее см. ниже).

* year: год точки данных.

* value: значение точки данных.

* age: код возрастной группы, к которой относится точка данных.

* pop: код единицы населения, к которой относится точка данных.

### Структура файла WID_metadata_XX.csv

Файл WID_metadata_XX.csv содержит семнадцать переменных:

* country: код страны/региона.

* variable: код переменной, на которую ссылаются метаданные.

* age: код возрастной группы, к которой относится население.

* pop: код единицы населения, к которой относится население.

* countryname: название страны/региона, как оно будет представлено в предложении на английском языке.

* shortname: название страны/региона, как оно будет представлено само по себе на английском языке.

* simpledes: описание переменной на простом английском языке.

* technicaldes: описание переменной с помощью учетных тождеств.

* shorttype: краткое описание типа переменной (среднее значение, агрегат, доля, индекс и т. д.) на простом английском языке.

* longtype: более подробное описание типа переменной на простом английском языке.

* shortpop: краткое описание единицы населения (индивидуумы, налоговые единицы, равные доли и т. д.) на простом английском языке.

* longpop: более подробное описание единицы населения на простом английском языке.

* shortage: краткое описание возрастной группы (взрослые, всё население и т. д.) простым языком.

* longage: более подробное описание возрастной группы простым языком.

* unit: единица измерения переменной (трёхбуквенный код валюты для денежных сумм).

* source: источник(и), использованный(е) для расчёта данных.

* method: методологические детали, описывающие, как были построены данные, и/или оговорки.

## Как интерпретировать коды переменных

Значение каждой переменной описано в файлах метаданных. Полные коды переменных WID (например, sptinc992j) подчиняются следующей логике:

* первая буква указывает тип переменной (например, «s» для доли).

* следующие пять букв указывают на доход/богатство/другое понятие (например, «ptinc» для национального дохода до вычета налогов).

* следующие три цифры указывают на возрастную группу (например, «992» для взрослых).

* последняя буква обозначает единицу населения (например, «j» для равного распределения).

## Как интерпретировать коды процентилей

На WID.world существует два типа процентилей: (1) групповые процентили и (2) обобщённые процентили. Интерпретация среднего, долевого или порогового ряда дохода (или богатства) зависит от типа рассматриваемого процентиля.

### Процентили группы

Процентили группы определяются следующим образом: p0p50 (нижние 50% населения), p50p90 (следующие 40%), p90p100 (верхние 10%), p99p100 (верхний 1%), p0p10 (нижние 10% населения, т.е. первый децил), p10p20 (следующие 10%, т.е. второй децил), p20p30 (следующие 10%, т.е. третий децил), p30p40 (следующие 10%, т.е. четвертый децил), p40p50 (следующие 10%, т.е. пятый децил), p50p60 (следующие 10%, т.е. шестой децил), p60p70 (следующие 10%, т.е. седьмой децил), p70p80 (следующие 10%, т.е. восьмой децил),

p80p90 (следующие 10%, т.е. девятый децил), p0p90 (нижние 90%), p0p99 (нижние 99% населения), p99,9p100 (верхние 0,1%), p99,99p100 (верхние 0,01%).

Для каждого процентиля группы мы приводим соответствующие доли дохода или богатства, средние значения и пороговые значения.

* Доли процентиля группы соответствуют доле дохода (или богатства), принадлежащей данному процентилю группы. Например, доля фискального дохода группы p0p50 — это доля общего фискального дохода, приходящаяся на нижние 50% группы.

* Средние значения процентиля группы соответствуют среднему годовому доходу (или богатству) в пределах данной процентильной группы. Например, средний фискальный доход группы p0p50 — это средний годовой фискальный доход нижней 50% группы.

* Пороговые значения группового процентиля соответствуют минимальному уровню дохода (или благосостояния), необходимому для принадлежности к данному процентилю. Например, пороговое значение налогового дохода для группы p90p100 — это минимальный годовой налоговый доход, необходимый для принадлежности к 10% самых богатых.

При наличии данных веб-сайт WID.world позволяет рассчитать доли, средние значения и пороговые значения для любого процентиля группы (например, средний доход 43p99,92p). Эти данные не хранятся в таблицах с данными.

Для некоторых стран из-за ограничений данных мы не можем предоставить список процентилей групп, описанных выше. Вместо этого мы сохраняем конкретные групповые процентили (в зависимости от страны это могут быть p90p95, p95p100, p95p99, p99.5p100, p99.5p99.9, p99.75p100, p99.95p100, p99.95p99.99, p99.995p100, p99.9p99.95, p99.9p99.99 или p99p99.5).

### Обобщённые процентили

Обобщённые процентили (g-процентили) определяются следующим образом: p0, p1, p2, ..., p99, p99.1, p99.2, ..., p99.9, p99.91, p99.92, ..., p99.99, p99.991, p99.992, ..., p99.999. Всего существует 127 g-процентилей.

Для каждого g-процентиля мы приводим доли, средние значения, максимальные средние значения и пороговые значения.

* Доли g-процентилей соответствуют доле дохода (или богатства), получаемой группой населения выше заданного значения g-процентиля. Например, доля фискального дохода g-процентиля p90 соответствует доле фискального дохода, принадлежащей 10% верхней группы; Доля фискального дохода в g-процентиле p99.9 соответствует доле фискального дохода верхней 0,1% группы и так далее. По построению, доля фискального дохода в g-процентиле p0 соответствует доле, принадлежащей 100% населения, и равна 100%. Формально, доля g-процентиля в g-процентиле pX соответствует доле верхней (100-X)% группы.

* Средние значения g-процентилей соответствуют среднему доходу или богатству между двумя последовательными g-процентилями. Средний доход g-процентиля p0 соответствует среднему годовому доходу нижней 1% группы, p2 соответствует следующей 1% группе и так далее до p98 (1% группы населения ниже верхней 1%). Средний доход g-процентиля p99 соответствует среднему годовому групповому процентилю p99p99.1 (т. е. 10% наименее обеспеченных внутри 1% наиболее обеспеченных), p99.1 соответствует следующей 0,1% группе, p99.2 соответствует следующей 0,1% группе и так далее до p99.8. Средний доход p99.9 соответствует среднему годовому доходу группового процентиля p99.9p99.91 (т. е. 10% наименее обеспеченных внутри 0,1% наиболее обеспеченных), p99.91 соответствует следующей 0,01% группе, p99.92 соответствует следующей 0,01% группе и так далее до p99.98. Средний доход p99,99 соответствует среднему годовому доходу процентиля группы p99,99p99,991 (т.е. 10% наименее обеспеченных граждан внутри 0,01% наиболее обеспеченных граждан), p99,991 соответствует следующим 0,001%, p99,992 соответствует следующим 0,001% и так далее до p99,999 (средний доход 0,001% наиболее обеспеченных граждан). Например, средний налоговый доход процентиля p50 равен среднему годовому налоговому доходу процентиля группы p50p51 (т.е. среднему годовому доходу группы населения, доход которой составляет более 50% населения и менее 49% наиболее обеспеченных граждан). Средний налоговый доход в g-процентиле p99 равен среднему годовому налоговому доходу в пределах процентиля группы p99p99.1 (т.е. группы, составляющей 0,1% от общей численности населения, с доходом более 99% населения, но менее 0,9% самых богатых).

* Верхние средние значения g-процентиля соответствуют среднему доходу или богатству выше заданного порогового значения g-процентиля. Например, верхний средний налоговый доход в g-процентиле p50 соответствует среднему годовому налоговому доходу лиц с доходом более 50% населения. Верхний средний налоговый доход в g-процентиле p90 соответствует среднему годовому налоговому доходу 10% самых богатых групп.

* Пороговые значения g-процентиля соответствуют минимальному уровню дохода (или богатства), необходимому для принадлежности к группе населения выше заданного значения g-процентиля. Например, пороговое значение налогового дохода в g-процентиле p90 соответствует минимальному годовому налоговому доходу, необходимому для принадлежности к 10% самым богатым группам. Порог фискального дохода в g-процентиле p99.9 соответствует минимальному годовому налоговому доходу, необходимому для принадлежности к верхней 0,1%-ной группе. Формально пороговое значение g-процентиля на уровне g-процентиля pX соответствует пороговому значению верхней (100-X)%-ной группы.
